```python
import matplotlib.pyplot as mp
import pandas as pd
import seaborn as sb
import numpy as np
```

```python
df_spark = spark.read.table("workspace.default.spx_with_indicators")

df = df_spark.toPandas()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
mp.hist(df['Close'], color='blue', ec='black', bins=15)
mp.xlabel('Close')
mp.ylabel('Price')
mp.title('Distribution of SPX Volume')
mp.show()
```

```python
mp.boxplot(df['Volume'])
mp.title("Volume Boxplot")
mp.show()
```

```python
x = df['SMA_200']
y = df['ret_1d']

mp.scatter(x, y, color='green', marker='o', s=100, alpha=0.7)
mp.xlabel('SMA_200')
mp.ylabel('ret_1d(%)')
mp.title('SMA_200 vs ret_1d')
mp.show()
```

```python
correlation_matrix = df.corr(numeric_only = True)

print (correlation_matrix)

sb.heatmap(correlation_matrix, cmap="YlGnBu", annot=False)

mp.show()
```
