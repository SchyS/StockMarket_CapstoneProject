```python
df = spark.read.format("delta").load(data_path)

df = df.withColumn("Date", F.to_date("Date"))
```

```python
#Trend filter: Close > SMA_200

#Entry: RSI < 40

#Exit: +2%, -2%, or 7 days

df_signal_close_SMA200 = (
    df.withColumn(
        "entry_signal",
        F.when(
            (F.col("Close") > F.col("SMA_200")) & 
            (F.col("RSI_14") < F.lit(40)),
            F.lit(1)
        ).otherwise(F.lit(0))   
    )
    #.withColumn(
        #"exit_signal",
        #F.when(F.col("SMA_20") < F.col("SMA_50"), F.lit(1)).otherwise(F.lit(0))
    #)
)
```

```python
#Entry: RSI < 40

#Exit: +2%, -2%, or 7 days

df_signal_RSI40 = (
    df.withColumn(
        "entry_signal",
        F.when(
            (F.col("RSI_14") < F.lit(40)),
            F.lit(1)
        ).otherwise(F.lit(0))   
    )
    #.withColumn(
        #"exit_signal",
        #F.when(F.col("SMA_20") < F.col("SMA_50"), F.lit(1)).otherwise(F.lit(0))
    #)
)
```

```python
#Here we are creating our entry signal to enter a long position based on the following conditions:
#- SMA_20 > SMA_50 
#- RSI_14 > 50

#Our exit signal is when the SMA_20 crosses below the SMA_50 
#SMA_20 <SMA_50

df_signal_SMA20_50 = (
    df.withColumn(
        "entry_signal",
        F.when(
            (F.col("SMA_20") > F.col("SMA_50")),
            F.lit(1)
        ).otherwise(F.lit(0))   
    )
    .withColumn(
        "exit_signal",
        F.when(F.col("SMA_20") < F.col("SMA_50"), F.lit(1)).otherwise(F.lit(0))
    )
)
```

```python
#Here we are creating our entry signal to enter a long position based on the following conditions:
#- SMA_20 > SMA_50 > SMA_200
#- RSI_14 > 50

#Our exit signal is when the SMA_20 crosses below the SMA_50 
#SMA_20 <SMA_50

df_signal_SMA20_50_200 = (
    df.withColumn(
        "entry_signal",
        F.when(
            (F.col("SMA_20") > F.col("SMA_50")) & 
            (F.col("SMA_50") > F.col("SMA_200")) &
            (F.col("RSI_14") > F.lit(50)),
            F.lit(1)
        ).otherwise(F.lit(0))   
    )
    .withColumn(
        "exit_signal",
        F.when(F.col("SMA_20") < F.col("SMA_50"), F.lit(1)).otherwise(F.lit(0))
    )
)
```

```python
trades_path = "dbfs:/Volumes/workspace/default/capstoneproject/S&P500_Data_Clean/trades_CloseSMA200"
```

```python
trades_schema = StructType([
    StructField("entry_date", DateType(), True),
    StructField("entry_price", DoubleType(), True),
    StructField("exit_date", DateType(), True),
    StructField("exit_price", DoubleType(), True),
    StructField("return", DoubleType(), True),
    StructField("mae", DoubleType(), True),
    StructField("reason", StringType(), True),
    StructField("pnl_price", DoubleType(), True)
])
```

```python
STOP_LOSS_PCT = 0.02
TAKE_PROFIT_PCT = 0.02
```

```python
def backtest_one_series(pdf):
    pdf = pdf.sort_values("Date").reset_index(drop=True)

    #Change amount of shares to simulate how much you want to invest
    shares = 1
    trades = []
    in_pos = False
    entry_price = None
    entry_date = None
    mae = 0.0

    i = 0
    n = len(pdf)

    while i < n - 1:
        row = pdf.iloc[i]

        if not in_pos:
            if int(row["entry_signal"]) == 1:
                entry_row = pdf.iloc[i + 1]
                in_pos = True
                entry_price = float(entry_row["Open"])
                entry_date = entry_row["Date"]
                mae = 0.0
                i = i + 1
                continue

        if in_pos:
            row = pdf.iloc[i]
            day_high = float(row["High"])
            day_low = float(row["Low"])
            day_close = float(row["Close"])

            mae = min(mae, (day_low - entry_price) / entry_price)

            stop_price = entry_price * (1 - STOP_LOSS_PCT)
            take_price = entry_price * (1 + TAKE_PROFIT_PCT)

            exit_reason = None
            exit_price = None
            exit_date = row["Date"]

            if day_low <= stop_price:
                exit_reason = "stop_loss"
                exit_price = stop_price

            elif day_high >= take_price:
                exit_reason = "take_profit"
                exit_price = take_price

            if exit_reason is not None:
                ret = (exit_price - entry_price) / entry_price
                pnl = (exit_price - entry_price) * shares
                trades.append({
                    "entry_date": entry_date,
                    "entry_price": entry_price,
                    "exit_date": exit_date,
                    "exit_price": exit_price,
                    "return": ret,
                    "mae": mae,
                    "reason": exit_reason,
                    "pnl_price": pnl
                })
                in_pos = False
                entry_price = None
                entry_date = None
                mae = 0.0

        i += 1
    return pd.DataFrame(trades)
                                              
```

```python
def createTable(df_signal, strat_name):
    df_one = df_signal.withColumn("series_id", F.lit("SPX"))

    trades_spark = (
        df_one
        .select(
            "series_id",
            "Date",
            "Open",
            "High",
            "Low",
            "Close",
            "SMA_20",
            "SMA_50",
            "SMA_200",
            "RSI_14",
            "entry_signal"
        )
        .groupBy("series_id")
        .applyInPandas(backtest_one_series, schema = trades_schema)
    )

    trades_spark.write.format("delta").mode("overwrite").option("overwriteSchema", True).save(trades_path)

    trades_spark.write.format("delta").mode("overwrite").option("overwriteSchema", True).saveAsTable(f"default.capstoneTrades_{strat_name}")
```
