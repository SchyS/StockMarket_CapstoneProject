#Model Comparisons

##Logistical Regression

The best overall strategy for the logistical regression was the trading strategy using the Simple Moving Average (SMA) 20 Days, SMA 50 and SMA 200. It had the highest accuracy at 0.71, the strongest F1 score at 0.71 and the best AUC out of all the other trading strategies at 0.92.

CloseSMA200 looks promising with the highest cross validation f1 score being 0.78 but it had 0 precision and recall. This means it essentially only predicted profitable trades. 

The second best seemed to be SMA20, SMA50 with an F1 score of 0.69 and an accuray of 0.59 but had a week recall at 0.3.

Finally the RSI40 trading strategy had lower F1 score 0.62 and accuracy of 0.55 but it had a higher auc of 0.8.

## Neural Network
The best performing strategy for the neural networks was the CloseSMA200 strategy. It had an accruacy of 0.8, and f1 score of 0.86 and an AUC of 0.83. 

The second best was the SMA20_50_200 trading strategy with an F1 score of 0.71, accuracy of 0.64 and an AUC = 0.65. 

Lastly the two weekend strategies were the SMA20_50 with an accuracy of 0.5 and an auc of 0.38 as well as the RSI40 with an accuracy of 0.36, F1 score of 0.46 and AUC of 0.4.

##Random Forest
For the random forest classifier SMA20_50_200 had the best accuracy of 0.71, F1 score of 0.75 and AUC of 0.81. This shows it had a strong and balanced performance across all of our metrics.

Next, the CloseSMA200 had an F1 score of 0.67, accuracy of 0.6 and an AUC of 0.5. Having an AUC of 0.5 is similiar to randomly guessing.

Further, we had the SMA20_50 with an F1 score of 0.62, an accuracy of 0.55 and an AUC of 0.58.

The worst performing overall was the RSI40 strategy with an accuracy of 0.36, F1 score of 0.46 and an AUC of 0.43 using this model.

## Conclusion
Model performance had varied across trading strategies but no one model stood out or outperformed the other (Logistic Regression, Neural Network or Random Forest). Neural Networks had the highest overall performance with the CloseSMA200 trading strategy, achieving an F1 Score of 0.86, showing its ability to capture nonlinear relationship. The Random Forest performed the best with the SMA20_50_200 strategy, with an F1 Score of 0.75, showing it could effectively model interactions between several technical indicators. Lastly, while Logistic Regression did not achieve the highest performances, it had the most consistent and stable results across trading strategies. These findings show the importance of matching our data characteristics with the correct model complexities; the more complex models such as our Neural Netowrk and Random Forests showed higher variance while the simpler models generalized more reliability. 

For trading strategies, two out of the three models, performed best on the SMA20_50_200 strategy. This strategy holds two layers while the others consist of one when determining when to enter the trade. Seeing this, an inference could be made stating that more specific, trialed and tested trading strategies could yield better results. Not only does this improve the chances of making profit or not, but also could improve machine models ability to help traders in their decisions to make trades.
